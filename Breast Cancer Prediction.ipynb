```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn.datasets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
#loading the data
breast_cancer_dataset = sklearn.datasets.load_breast_cancer()
```

```python
print(breast_cancer_dataset)
```

```python
#loading the dataset into pandas dataframe
df = pd.DataFrame(breast_cancer_dataset.data, columns = breast_cancer_dataset.feature_names)
```

```python
df.head()
```

```python
df['label'] = breast_cancer_dataset.target
```

```python
#separating features and target
X = df.drop('label', axis = 1)
Y = df['label']
```

```python
print(X)
```

```python
#splitting the data
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, stratify = Y, random_state = 2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

```python
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

**Building Neural Network**

```python
#importing tensorflow and keras
import tensorflow as tf
tf.random.set_seed(3) #3 is just a random number to reproduce a particular result like random_state in train_test_split
from tensorflow import keras
```

```python
#setting up my neural network layers
model = keras.Sequential([
    keras.layers.Flatten(input_shape = (30,)), #30 is the number of features
    keras.layers.Dense(20, activation = 'relu'), #20 is number of neurons for hidden layer
    keras.layers.Dense(2, activation = 'softmax') #2 is number of neurons for output layer == number of classes i.e 0 & 1
])
```

```python
#compiling the Neural Ntwork
model.compile(
    optimizer = 'adam', # How the model learns
    loss = 'sparse_categorical_crossentropy', # How errors are measured
    metrics = ['accuracy']
)
```

```python
#training the Neural Network
history = model.fit(X_train, Y_train, validation_split = 0.1, epochs = 10) #0.1 is 10% validation split
```

Visualizing Accuracy and loss

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('Model accuracy')
plt.ylabel('accuracy')
plt.xlabel('epoch')
plt.legend(['training data', 'validation data'], loc = 'lower right')
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('Model loss')
plt.ylabel('loss')
plt.xlabel('epoch')
plt.legend(['training data', 'validation data'], loc = 'upper right')
```

Accuracy of model on test data

```python
loss, accuracy = model.evaluate(X_test, Y_test)
print(accuracy)
```
